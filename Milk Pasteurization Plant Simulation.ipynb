```python
import time

# =============================================================================
# SUB-PROCESS 1: STORAGE TANK CONTROL
# =============================================================================
def control_storage_tank(milk_temperature, milk_level_sensor):
    print("\n--- [STAGE 1] STORAGE TANK STATUS ---")

    # Thermal Regulation Loop
    if milk_temperature > 4:
        print(f"Warning: Milk temperature is {milk_temperature}°C (Higher than 4°C).")
        print("-> Status: COOLING SYSTEM ACTIVATED to bring it down.")
    else:
        print(f"Status: Milk temperature is safe at {milk_temperature}°C. Cooling system is idle.")

    # Safety Interlock Loop for Agitator
    if milk_level_sensor == 1:
        print("Status: Milk level is safe (Level Sensor = 1).")
        print("-> Propeller Motor: ON (Keeping milk uniform and homogeneous).")
    else:
        print("Warning: Milk level is too low (Level Sensor = 0).")
        print("-> Propeller Motor: OFF (Stopped to prevent splashing and foaming).")


# =============================================================================
# SUB-PROCESS 2: BALANCE TANK CONTROL
# =============================================================================
def control_balance_tank(balance_tank_level_sensor):
    print("\n--- [STAGE 2] BALANCE TANK STATUS ---")

    # Regulating the inlet valve
    if balance_tank_level_sensor == 1:
        print("Status: Balance tank is full (Sensor = 1).")
        print("-> Inlet Valve: CLOSED (Preventing overflow).")
    else:
        print("Status: Balance tank level is low (Sensor = 0).")
        print("-> Inlet Valve: OPEN (Refilling to maintain constant flow).")


# =============================================================================
# SUB-PROCESS 3: HEATER CONTROL
# =============================================================================
def control_heater(milk_flow_rate, milk_inlet_temperature, water_inlet_temperature, water_outlet_temperature):
    print("\n--- [STAGE 3] HEATER VALVE REGULATION ---")

    # Specific heat capacities given in the document
    specific_heat_capacity_of_milk = 3.89   # kJ/kg
    specific_heat_capacity_of_water = 4.18  # kJ/kg
    target_milk_temperature = 72            # Required pasteurization temperature in °C

    # Formula from Equation 01
    numerator_value = milk_flow_rate * specific_heat_capacity_of_milk * (target_milk_temperature - milk_inlet_temperature)
    denominator_value = specific_heat_capacity_of_water * (water_inlet_temperature - water_outlet_temperature)

    # Avoiding division by zero error
        if denominator_value == 0:
        print("Error: Water inlet and water outlet temperatures cannot be identical.")
        return 0

    water_flow_rate_required = numerator_value / denominator_value

    print(f"Target Milk Temperature: {target_milk_temperature}°C")
    print(f"Calculated Hot Water Flow Rate needed: {water_flow_rate_required:.2f} kg/s")
    print("-> Status: Heating valve adjusted to match the calculated flow rate.")
    return target_milk_temperature


# =============================================================================
# SUB-PROCESS 4: HOLDING TUBE
# =============================================================================
def run_holding_tube(current_milk_temperature):
    print("\n--- [STAGE 4] HOLDING TUBE TIMING ---")

    # Check if milk is at the correct pasteurization temperature
    if current_milk_temperature >= 72:
        print("Success: Milk reached 72°C. Starting 15-second pathogen elimination countdown...")

        # Simulating a simple countdown loop for 15 seconds
        for holding_timer in range(1, 16):
            print(f"Holding... Time elapsed: {holding_timer} second(s)")
            time.sleep(0.1)  # Simulation

        print("-> Complete: Milk has been safely pasteurized for 15 seconds.")
        return True
    else:
        print("Safety Alert: Milk temperature is below 72°C! Diverting flow back to heater.")
        return False


# =============================================================================
# SUB-PROCESS 5: COOLING CONTROL
# =============================================================================
def run_cooler():
    print("\n--- [STAGE 5] COOLING PROCESS ---")
    print("Status: Pasteurized hot milk entering the cooling chamber.")
    print("Cooling Medium: Chilled water entering at 1°C.")

    # Rapid cooling simulation loop
    cooled_milk_temperature = 72
    while cooled_milk_temperature > 4:
        cooled_milk_temperature -= 17  # Rapid thermal quench simulation step
        if cooled_milk_temperature < 4:
            cooled_milk_temperature = 4
        print(f"Cooling down... Current milk temp: {cooled_milk_temperature}°C")

    print("-> Ready: Milk successfully cooled down to safe storage temperature of 4°C.")
    print("-> Status: Forwarding to Packager stage for bottling.")


# =============================================================================
# MAIN PROGRAM EXECUTION (SIMULATION START)
# =============================================================================
print("==================================================")
print("     MILK PASTEURIZATION PLANT SIMULATION         ")
print("==================================================")

# 1. Storage Tank Control (Milk Temp: 5°C, Level Sensor: 1)
control_storage_tank(milk_temperature=5, milk_level_sensor=1)

# 2. Balance Tank Control (Level Sensor: 0 [Needs refill])
control_balance_tank(balance_tank_level_sensor=0)

# 3. Heater Control
milk_outlet_temperature = control_heater(
    milk_flow_rate=2.5,
    milk_inlet_temperature=4,
    water_inlet_temperature=85,
    water_outlet_temperature=75
)

# 4. Holding Tube Stage
is_pasteurization_successful = run_holding_tube(milk_outlet_temperature)

# 5. Cooling & Packaging Stage
if is_pasteurization_successful:
    run_cooler()
    print("\n--- [STAGE 6] PACKAGING ---")
    print("Status: Milk packed into bottles/cartons successfully.")

print("\n==================================================")
print("     PROCESS COMPLETED SUCCESSFULLY!")
print("==================================================")
```
